```python
import torch
from torch import nn

torch.device('cpu'), torch.cuda.device('cuda:0') # 这台设备没有gpu
```

查询可用GPU的数量

```python
torch.cuda.device_count()
```

在请求的GPU不存在的情况下运行代码

```python
def try_gpu(i):
    if torch.cuda.device_count() >= i + 1:
        return torch.cuda.device(f'cuda:{i}')
    else:
        return torch.device('cpu')
x = torch.tensor([0], device = try_gpu(0))
x.device
```

```python
def try_all_gpu():
    devices = [torch.cuda.device(f'cuda:{i}') for i in range(torch.cuda.device_count())]
    return devices if devices else [torch.device('cpu')]
try_gpu(0), try_all_gpu()
```
