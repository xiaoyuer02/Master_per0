# 填充和步幅

在所有侧边填充1个像素

```python
import torch
from torch import nn

net = nn.Conv2d(1, 1, 3, padding = 1)
X = torch.ones((8, 8))
def comp_conv2d(net, X):
    X = X.reshape((1, 1) + X.shape)
    Y = net(X)
    Y = Y.reshape(Y.shape[2:])
    return Y
Y = comp_conv2d(net, X)
Y.shape # 输出和输入维度没变
```

填充不同的高度和宽度

```python
net1 = nn.Conv2d(1, 1, kernel_size = (5, 3), padding = (2, 1))
Y = comp_conv2d(net1, X)
Y.shape
```

将高度和宽度的步幅设置为2

```python
net2 = nn.Conv2d(1, 1, kernel_size = 3, padding = 1, stride = 2)
Y = comp_conv2d(net2, X)
Y.shape
```

更复杂的例子

```python
net3 = nn.Conv2d(1, 1, kernel_size = (3, 5), padding = (0, 1), stride = (3, 4))
Y = comp_conv2d(net3, X)
Y.shape # [(n+padding*2-kernel_size)/stride +1]
```
