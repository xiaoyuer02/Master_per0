加载或保存张量

```python
import torch
from torch import nn
from torch.nn import functional as F

x = torch.arange(12)
torch.save(x, 'x-file.pt')
x2 = torch.load('x-file.pt')
x2 == x
```

存储一个张量列表

```python
y = torch.arange(4)
torch.save([x, y], 'xy-files.pt')
x2, y2 = torch.load('xy-files.pt')
(x2, y2)
```

写入或读取从字符串映射到张量的字典

```python
dict = {'x': x, 'y': y}
torch.save(dict, 'xy-dict.pt')
dict2 = torch.load('xy-dict.pt')
dict2
```

加载或保存模型参数

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(20, 256)
        self.output = nn.Linear(256, 10)

    def forward(self, X):
        return self.output(self.linear(X))
net = MLP()
X = torch.rand((2, 20))
Y = net(X)
```

```python
torch.save(net.state_dict(), 'mlp-params.pt') # state_dict()函数返回参数有序字典 
```

```python
clone = MLP() # 先创建一个实例
clone.load_state_dict(torch.load('mlp-params.pt')) # 加载state_dict后再写入覆盖load_state_dict()
clone.eval() # 切评估模式 会影响Dropout层和BatchNorm层 使得每次评估都一致 但对F.dropout需要另传training = False关闭
```

```python
Y == clone(X)
```
